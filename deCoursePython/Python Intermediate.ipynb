```python
def database_columns(*args):
    for arg in args:
        print(arg)

database_columns('name', 'age', 'height')
```

```python
# **kwargs is for kvps like dictionary
def database_column(**kwargs):
    for value in kwargs.items():
        print(value)

database_column(name='Darshil', age=25)
```
