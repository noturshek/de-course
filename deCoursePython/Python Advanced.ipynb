# Pandas Data Strucutres:
1. Series:
   * One-dimensional labeled array
   * Supports Multiple Data Types
2. Data Frame:
   * Two-dimensional labeled array
   * Supports Multiple Data Types
   * Input can be series
   * Input can be other data frame
3. Panel:
   * Three-dimensional labled array
   * Supports multiple data types
   * Item-> axis 0
   * Major axis -> rows
   * Minor axis -> columns
4. Panel 4D(experimental):
   * Four-dimensional labeled array
   * Supports multiple data types
   * Labels -> axis 0
   * Items -> axis 1
   * Major axis -> rows
   * Minor axis -> columns

# Pandas Series:

```python
import pandas as pd

my_list = [10,20,30]
ser = pd.Series(my_list)

print(ser)
print(ser.index)
print(ser.values)
```

```python
# Creating a series from numpy array
import numpy as np
import pandas as pd

index = ['a','b','c']
arr = np.array([10,20,30])

pd.Series(data=arr, index=index)
```

```python
# Creating a series from dictionary
import pandas as pd

d= {'a':10, 'b':20, 'c':30}
pd.Series(d)
```
