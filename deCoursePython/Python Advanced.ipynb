# Pandas Data Strucutres:
1. Series:
   * One-dimensional labeled array
   * Supports Multiple Data Types
2. Data Frame:
   * Two-dimensional labeled array
   * Supports Multiple Data Types
   * Input can be series
   * Input can be other data frame
3. Panel:
   * Three-dimensional labled array
   * Supports multiple data types
   * Item-> axis 0
   * Major axis -> rows
   * Minor axis -> columns
4. Panel 4D(experimental):
   * Four-dimensional labeled array
   * Supports multiple data types
   * Labels -> axis 0
   * Items -> axis 1
   * Major axis -> rows
   * Minor axis -> columns

```python
# Creating a series from numpy array
import numpy as np
import pandas as pd

index = ['a','b','c']
arr = np.array([10,20,30])

pd.Series(data=arr, index=index)
```

# Series Operations

```python
weather_df['Wind Spd (km/h)'].head()
```

```python
weather_df['Wind Spd (km/h)'].head() + 10
```

```python
weather_df.head()
```

```python
weather_df['new_temp_col'] = weather_df['Temp (C)'] + weather_df['Dew Point Temp (C)']
weather_df.head()
```

```python
# .apply : understand from the example

def times2(value):
    return value * 2

t2 = weather_df['Visibility (km)'].apply(times2)
t2.head()
```

```python
weather_df['Visibility (km)'].describe()
```

```python
# Adding/Updating columns
visibility_in_meter = weather_df['Visibility (km)'] * 1000
weather_df['Visibility (m)'] = visibility_in_meter

weather_df.head()
```

```python
weather_df.rename(columns={'Visibility (m)':'Visibility (meters)',
                          'Visibility (km)':'Visibility (kilometers)'}, inplace = True)
weather_df.head()
```

```python
weather_df.rename(columns={'Visibility (kilometers)':'Visibility (km)'}, inplace=True)
weather_df.head()
```

```python
weather_df.drop(labels = ['Visibility (meters)'], axis=1)
weather_df.head()
```

```python
# Sorting
sorted_by_temp = weather_df.sort_values('Temp (C)', ascending=False)
sorted_by_temp.head()
```

```python
weather_df = weather_df.drop(['Visibility (meters)'], axis = 1)
weather_df.head()
```

```python
# Which were the top 10 hottest values and their count
sorted_value_count = weather_df['Temp (C)'].value_counts().sort_values(ascending = False)
sorted_value_count
```
