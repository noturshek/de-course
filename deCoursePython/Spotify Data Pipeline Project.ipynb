```python
import spotipy
from spotipy import Spotify
from spotipy.oauth2 import SpotifyOAuth
from dotenv import load_dotenv
import os

load_dotenv()
CLIENT_ID = os.getenv("SPOTIFY_CLIENT_ID")
CLIENT_SECRET = os.getenv("SPOTIFY_CLIENT_SECRET")
REDIRECT_URI = os.getenv("SPOTIFY_REDIRECT_URI")
```

```python
sp = Spotify(
    auth_manager=SpotifyOAuth(
        client_id=CLIENT_ID,
        client_secret=CLIENT_SECRET,
        redirect_uri=REDIRECT_URI,
        scope='playlist-read-private playlist-read-collaborative',
        open_browser=True
    )
)
```

```python
playlist_link = 'https://open.spotify.com/playlist/1npQQ8nEAzp256gIK6Mf1d'
playlist_URI = playlist_link.split('/')[-1]
```

```python
data = sp.playlist_tracks(playlist_URI)
```

```python
album_list = []

for item in data['items']:
    row = item['item']['album']

    album_id = row['id']
    album_name = row['name']
    album_release_date = row['release_date']
    album_total_tracks = row['total_tracks']
    album_url = row['external_urls']['spotify']
    album_artists = ", ".join(artist["name"] for artist in row["artists"])

    album_element = {
        'album_id': album_id,
        'name': album_name,
        'release_date': album_release_date,
        'total_tracks': album_total_tracks,
        'url': album_url,
        'artist_name':album_artists
    }

    album_list.append(album_element)

album_list
```

```python
data['items'][0]['item']['album']['artists'][0]
```

```python
artist_list = []
for row in data['items']:
    artists = row['item']['artists']

    for artist in artists:
        artist_id = artist['id']
        artist_name = artist['name']
        artist_url = artist['href']

        artist_element = {
            'artist_id': artist_id,
            'artist_name': artist_name,
            'external_url': artist_url
        }

        artist_list.append(artist_element)

artist_list
```

```python
import random
song_list = []

for row in data['items']:
    song = row['item']

    song_id = song['id']
    song_name = song['name']
    song_duration = song['duration_ms']
    song_url = song['external_urls']['spotify']
    song_added = row['added_at']
    album_id = song['album']['id']
    artist_id = ", ".join(
        artist['id'] for artist in song['artists']
    )

    song_element = {
        'song_id': song_id,
        'song_name': song_name,
        'duration_ms': song_duration,
        'url': song_url,
        'song_added': song_added,
        'album_id': album_id,
        'artist_id': artist_id,
        'song_popularity': random.randint(75, 95)
    }

    song_list.append(song_element)

song_list
```

```python
import pandas as pd

album_df = pd.DataFrame.from_dict(album_list)
album_df
```

```python
album_df.info()
```

```python
album_df = album_df.drop_duplicates(subset=['album_id'])
```

```python
artist_df = pd.DataFrame.from_dict(artist_list)
artist_df
```

```python
artist_df = artist_df.drop_duplicates(subset=['artist_id'])
```

```python
song_df = pd.DataFrame.from_dict(song_list)
song_df
```

```python
song_df = song_df.drop_duplicates(subset=['song_id'])
```
