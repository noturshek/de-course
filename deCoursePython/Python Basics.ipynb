```python
# Print Formatting
name = 'Abhishek Anant Singh'
age = 24
married = False

print("My name is %s aged %s and it is a %s statement that I am married" % (name, age, married))
print("My name is {} aged {} and it is a {} statement that I am married".format(name, age, married))
print(f"My name is {name} aged {age} and it is a {married} statement that I am married")
```
